In this laboratory exercise, we will explore the concepts of souce coding. We will focus on Huffman coding. Huffman codes are uniquely decodable instantaneous codes with minimum-average code word length. In this sense they are optimal. By optimality we  mean that, among all codes that satisfy the prefix condition (and therefore are uniquely  decodable and instantaneous), Huffman codes have the minimum-average code word  length.

### Step 1: Generation of a random alphabet

```python
import random
from heapq import heappush, heappop, heapify# Generate random probabilities for letters a-p
letters = [chr(i) for i in range(ord('a'), ord('p')+1)]
probs = [random.random() for _ in letters]
probabilities = {letters[i]: round(probs[i] / sum(probs), 8) for i in range(len(letters))}# Build the Huffman tree
heap = [[p, [l, ""]] for l, p in probabilities.items()]
heapify(heap)
while len(heap) > 1:
    lo, hi = heappop(heap), heappop(heap)
    for pair in lo[1:]: pair[1] = '0' + pair[1]
    for pair in hi[1:]: pair[1] = '1' + pair[1]
    heappush(heap, [lo[0] + hi[0]] + lo[1:] + hi[1:])
# Output results
print("Probabilities:", probabilities)
print("Huffman Codes:", {sym: code for sym, code in sorted(heappop(heap)[1:], key=lambda x: x[0])})
```

```python
import heapq
# Step 1: Define the probabilities for each symbol
heap = [[weight, [symbol, ""]] for symbol, weight in probabilities.items()]#Step 2: Create a heap where each element is [probability, [symbol, ""]]
heapq.heapify(heap)  # Convert the list into a min-heap
while len(heap) > 1:# Step 3: Build the Huffman tree by combining the two nodes with the smallest probabilities
    lo = heapq.heappop(heap)# Pop the two nodes with the smallest probabilities
    hi = heapq.heappop(heap)# Add '0' to the code of the first node and '1' to the code of the second node
    for pair in lo[1:]:
        pair[1] = '0' + pair[1]  # Update code for the first symbol
    for pair in hi[1:]:
        pair[1] = '1' + pair[1]  # Update code for the second symbol
    heapq.heappush(heap, [lo[0] + hi[0]] + lo[1:] + hi[1:])# Merge the two nodes and push the merged node back into the heap
# Step 4: Get the final Huffman codes (after all nodes are merged)
codes = sorted(heapq.heappop(heap)[1:], key=lambda p: (len(p[1]), p))# Step 5: Convert the list of codes to a dictionary for easy lookup
code_dict = {item[0]: item[1] for item in codes}
for letter, code in code_dict.items():# Step 6: Print each symbol and its corresponding Huffman code
    print(f"{letter}: {code}")
```
