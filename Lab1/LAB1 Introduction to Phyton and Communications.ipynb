# Student task: Work with loops Display all prime numbers within an interval of 0 to 50.

```python
# Define the range
start = 2
to = 50
# Start with the first number in the range
num = start
# Loop through numbers until we reach the end of the range
while num <= to:
    # Assume the number is prime until proven otherwise
    is_prime = True
        # Check for factors from 2 up to the square root of num
    i = 2
    while i * i <= num:
        if num % i == 0:  # If num is divisible by i, it's not prime
            is_prime = False
            break
        i += 1
        # If no divisors were found, num is prime
    if is_prime:
        print(num, end=" ")
    
    # Move to the next number
    num += 1
```

# Student task: Work with matrices

# A:Create two matrices of dimensions 2x2.
# B:Transpose your matrices.
# C: Define a function to multiply your matrices.

```python
# Step 1: Define two 2x2 matrices
matrixA = [[1, 2], [3, 4]]
matrixB = [[5, 6], [7, 8]]

# Step 2: Transpose function for a 2x2 matrix
def transpose(matrix):
    return [[matrix[0][0], matrix[1][0]], [matrix[0][1], matrix[1][1]]]

# Transpose the matrices
transpose_matrixA = transpose(matrixA)
transpose_matrixB = transpose(matrixB)
print("Transpose of Matrix 1:", transpose_matrixA)
print("Transpose of Matrix 2:", transpose_matrixB)

# Step 3: Matrix multiplication function for two 2x2 matrices

def procedure_multiply_matrices(mat1, mat2):
    # Calculate each element of the resulting matrix individually
    a = mat1[0][0] * mat2[0][0] + mat1[0][1] * mat2[1][0]
    b = mat1[0][0] * mat2[0][1] + mat1[0][1] * mat2[1][1]
    c = mat1[1][0] * mat2[0][0] + mat1[1][1] * mat2[1][0]
    d = mat1[1][0] * mat2[0][1] + mat1[1][1] * mat2[1][1]
    
    # Return the result as a 2x2 matrix
    return [[a, b], [c, d]]


matrixA = [[1, 2], [3, 4]]
matrixB = [[5, 6], [7, 8]]

product_matrix = procedure_multiply_matrices(matrixA, matrixB)
print("Product of Matrix 1 and Matrix 2:", product_matrix)
```

# Generating a complex signal


The code below demonstrates how to generate a complex signal.

```python
#Example:
import numpy as np
import matplotlib.pyplot as plt
# Complex exponential signal
complex_exp_signal = np.exp(1j * t)  # 1j 

# Create a time vector
t = np.linspace(0, 2*np.pi, 500)

# Calculate complex exponential signal
complex_exp_signal = np.exp(1j * t)

# Draw a complex plane graph
plt.figure(figsize=(14, 6))

# Subfigure 1: Complex plane representation
plt.subplot(1, 2, 1)
plt.plot(np.real(complex_exp_signal), np.imag(complex_exp_signal)) 
plt.title('Complex Exponential Signal on Complex Plane')
plt.xlabel('Real Part')
plt.ylabel('Imaginary Part')
plt.grid(True)
plt.axis('equal')

# Sub-figure 2: Real and imaginary parts changing over time
plt.subplot(1, 2, 2)
plt.plot(t, np.real(complex_exp_signal), label='Real Part')
plt.plot(t, np.imag(complex_exp_signal), label='Imaginary Part', linestyle='--')
plt.title('Real and Imaginary Parts over Time')
plt.xlabel('Time')
plt.ylabel('Amplitude')
plt.grid(True)
plt.legend()

# Display the graph
plt.tight_layout()
plt.show()
```

# Student task: Work with complex signals and plots

# Based on the previous example, generate a complex signal consisting of two frequency components. The first frequency component should be $f_1 = 1000$ Hz; the second frequency component should be $f_2 = 10 $ Hz. Plot the curve showing how the amplitude of the signal you generate varies with time.

```python
import numpy as np
import matplotlib.pyplot as plt

# Define parameters
f1 = 1000  # Frequency component 1 in Hz
f2 = 10    # Frequency component 2 in Hz
sampling_rate = 10000  # Sampling rate to capture high-frequency components
duration = 0.01  # Duration of the signal in seconds

# Time vector
t = np.linspace(0, duration, int(sampling_rate * duration))

# Generate the complex signal with two frequency components
complex_signal = np.exp(1j * 2 * np.pi * f1 * t) + np.exp(1j * 2 * np.pi * f2 * t)

# Calculate the amplitude of the signal (magnitude)
amplitude = np.abs(complex_signal)

# Plot the amplitude of the signal over time
plt.figure(figsize=(10, 4))
plt.plot(t, amplitude)
plt.title("Amplitude of Complex Signal with Two Frequency Components")
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()
```

# Student task: Work with random variables and probability functions   


# Based on the previous example, generate a probability density function of exponential distribution (scale parameter is 1, that is, λ=1). Plot the PDF.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import expon

# Set the range of x (for an exponential distribution, the range is [0, ∞), so we use positive x)
x = np.linspace(0, 4, 1000)

# Probability density function of exponential distribution (scale = 1, λ = 1)
pdf_expon = expon.pdf(x, scale=1)
mean_expon = expon.mean(scale=1)
var_expon = expon.var(scale=1)

# Plot the exponential distribution
plt.figure(figsize=(12, 6))
plt.plot(x, pdf_expon, label='PDF Exponential', color='orange')
plt.title('Exponential Distribution (λ=1)')
plt.xlabel('x')
plt.ylabel('Density')
plt.grid(True)
plt.legend()

# Display the mean and variance on the plot
plt.text(1.5, 0.6, f'Mean: {mean_expon:.2f}\nVariance: {var_expon:.2f}', color='black')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the exponential distribution PDF with λ = 1
def exponential_pdf(x, lam=1):
    return lam * np.exp(-lam * x)

# Set the range of x (non-negative values for the exponential distribution)
x = np.linspace(0, 4, 1000)

# Compute the PDF values using the custom function
pdf_exponential = exponential_pdf(x)

# Plot the exponential distribution
plt.figure(figsize=(12, 6))
plt.plot(x, pdf_exponential, label='PDF Exponential (λ=1)', color='green')
plt.title('Exponential Distribution (λ=1)')
plt.xlabel('x')
plt.ylabel('Density')
plt.grid(True)
plt.legend()

# Calculate mean and variance analytically for λ = 1
mean_expon = 1 / 1  # mean = 1 / λ
var_expon = 1 / (1 ** 2)  # variance = 1 / λ^2

# Display the mean and variance on the plot
plt.text(2, 0.6, f'Mean: {mean_expon:.2f}\nVariance: {var_expon:.2f}', color='black')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import find_peaks
from scipy.fftpack import fft, fftfreq

# Define the exponential signal
def exponential_signal(x, lam=1):
    return lam * np.exp(-lam * x)

# Set the range of x
x = np.linspace(0, 4, 1000)

# Generate the exponential signal
signal = exponential_signal(x)

# Perform FFT to get frequency components
signal_fft = fft(signal)
frequencies = fftfreq(len(signal), d=(x[1] - x[0]))  # Frequency values

# Find peaks in the signal using scipy.signal.find_peaks (just for demo purposes)
peaks, _ = find_peaks(signal)

# Plot the original exponential signal
plt.figure(figsize=(14, 6))

# Subfigure 1: Exponential signal (time domain)
plt.subplot(1, 2, 1)
plt.plot(x, signal, label='Exponential Signal (λ=1)', color='green')
plt.plot(x[peaks], signal[peaks], 'rx', label='Peaks')
plt.title('Exponential Signal (Time Domain)')
plt.xlabel('x')
plt.ylabel('Amplitude')
plt.legend()
plt.grid(True)

# Subfigure 2: Fourier Transform (frequency domain)
plt.subplot(1, 2, 2)
plt.plot(frequencies, np.abs(signal_fft), color='blue')
plt.title('FFT of Exponential Signal')
plt.xlabel('Frequency')
plt.ylabel('Amplitude')
plt.grid(True)

# Display the plots
plt.tight_layout()
plt.show()
```
